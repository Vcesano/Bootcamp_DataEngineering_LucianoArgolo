```python
from pyspark.sql.functions import count, avg, min, max, col, round, desc
df = spark.table("bootcamp_de_valentin.gold.fact_propiedades")
df.groupBy("zona_id").agg(
    count("*").alias("cantidad"),
    round(avg("precio"),2).alias("precio_promedio"),
    min(col("precio")).alias("min_precio"),
    max(col("precio")).alias("max_precio"),
).orderBy(desc("cantidad")).explain(True)
```

```python
# ejercicio 3.2
# join
df_zona = spark.table("bootcamp_de_valentin.gold.dim_zona")
df_to = spark.table("bootcamp_de_valentin.gold.dim_tipo_operacion")
df_join =(
    df
    .join(df_zona, df.zona_id == df_zona.zona_id, "inner")
    .join(df_to, df.tipo_operacion_id == df_to.tipo_operacion_id, "inner")
    .select(
        df_zona.partido,
        df_zona.region,
        df.precio,
        df.precio_m2,
        df_to.moneda
    )
)
df_join.show(10)

```

```python
# ejercicio 3.3
from pyspark.sql.functions import count, avg, min, max, col, round, desc
df_estado = spark.table("bootcamp_de_valentin.gold.dim_caracteristicas")
df = spark.table("bootcamp_de_valentin.gold.fact_propiedades")
df_join = (
    df
    .join(df_estado, df.caracteristicas_id == df_estado.caracteristica_id, "inner")
    .join(df_zona, df.zona_id == df_zona.zona_id, "inner")
    .join(df_to, df.tipo_operacion_id == df_to.tipo_operacion_id, "inner")
    .groupBy(df_zona.partido, df_to.tipo_operacion)
    .agg(
        count("*").alias("conteo"),
        avg(df.precio).alias("precio_promedio"),
        avg(df.precio_m2).alias("precio_m2_promedio")
    )
    .filter(col("conteo")>50)
    .orderBy(desc("precio_promedio"))
)
df_join.show()
```

```python
# ejercicio 3.4
# window functions
from pyspark.sql.window import Window
from pyspark.sql.functions import count, avg, min, max, col, round, desc, row_number, rank, lag, abs
df = spark.table("bootcamp_de_valentin.gold.fact_propiedades")
df_zona = spark.table("bootcamp_de_valentin.gold.dim_zona")
df_to = spark.table("bootcamp_de_valentin.gold.dim_tipo_operacion")
df_join =(
    df
    .join(df_zona, df.zona_id == df_zona.zona_id, "inner")
    .join(df_to, df.tipo_operacion_id == df_to.tipo_operacion_id, "inner")
    .filter(col("moneda") == "USD")
    .filter(col("tipo_operacion") == "venta")
    .select("partido", "url", "precio", "precio_m2")
)

window = Window.partitionBy("partido").orderBy(desc("precio"))

df_ranking =(
    df_join
    .withColumn("row_num", row_number().over(window))
    .withColumn("rank", rank().over(window))
    .filter(col("row_num")<=3)
    .orderBy("partido","row_num")
)

df_ranking.show(30)
```

```python
# ejercicio 3.5
# lag
df = spark.table("bootcamp_de_valentin.gold.fact_propiedades")
df_zona = spark.table("bootcamp_de_valentin.gold.dim_zona")
df_to = spark.table("bootcamp_de_valentin.gold.dim_tipo_operacion")
df_join =(
    df
    .join(df_zona, df.zona_id == df_zona.zona_id, "inner")
    .join(df_to, df.tipo_operacion_id == df_to.tipo_operacion_id, "inner")
    .filter(df_to.moneda == "USD")
    .filter(df_to.tipo_operacion == "venta")
    .select(df_zona.partido, df.url, df.precio, df.precio_m2)
)

window = Window.partitionBy("partido").orderBy(desc("precio"))

df_lag = (
    df_join
    .withColumn("precio_siguiente", lag("precio",1).over(window))
    .withColumn("diferencia", col("precio") - col("precio_siguiente"))
    .filter(col("precio_siguiente").isNotNull())
    .orderBy("partido",desc(abs(col("diferencia"))))
)
df_lag.show(30)
```

```python
# ejercicio 3.6
# persistir DELTA
from pyspark.sql.functions import count, avg, min, max, col, round, desc
df_estado = spark.table("bootcamp_de_valentin.gold.dim_caracteristicas")
df = spark.table("bootcamp_de_valentin.gold.fact_propiedades")
df_zona = spark.table("bootcamp_de_valentin.gold.dim_zona")
df_to = spark.table("bootcamp_de_valentin.gold.dim_tipo_operacion")
df_join = (
    df
    .join(df_estado, df.caracteristicas_id == df_estado.caracteristica_id, "inner")
    .join(df_zona, df.zona_id == df_zona.zona_id, "inner")
    .join(df_to, df.tipo_operacion_id == df_to.tipo_operacion_id, "inner")
    .groupBy(df_zona.partido, df_to.tipo_operacion)
    .agg(
        count("*").alias("conteo"),
        avg(df.precio).alias("precio_promedio"),
        avg(df.precio_m2).alias("precio_m2_promedio")
    )
    .filter(col("conteo")>50)
    .orderBy(desc("precio_promedio"))
)

df_join.write.format("delta").mode("overwrite").saveAsTable("bootcamp_de_valentin.gold.resumen_zona_tipo_s6")
```

```python
# ejercicio 3.7
# spark.sql
query_spark_sql = spark.sql(f""" 
                  
                    SELECT
                        dz.partido,
                        dto.tipo_operacion,
                        count(*) `conteo`,
                        avg(fp.precio) `precio_promedio`,
                        avg(fp.precio_m2) `precio_m2_promedio`
                    FROM bootcamp_de_valentin.gold.fact_propiedades fp
                    inner join bootcamp_de_valentin.gold.dim_zona dz on fp.zona_id = dz.zona_id
                    inner join bootcamp_de_valentin.gold.dim_tipo_operacion dto on fp.tipo_operacion_id = dto.tipo_operacion_id
                    inner join bootcamp_de_valentin.gold.dim_caracteristicas dc on fp.caracteristicas_id = dc.caracteristica_id
                    group by dz.partido, dto.tipo_operacion
                    HAVING conteo > 50
                    ORDER BY precio_promedio desc

                  """
)
```

```python
# dataframe API
from pyspark.sql.functions import count, avg, min, max, col, round, desc
df_estado = spark.table("bootcamp_de_valentin.gold.dim_caracteristicas")
df = spark.table("bootcamp_de_valentin.gold.fact_propiedades")
df_zona = spark.table("bootcamp_de_valentin.gold.dim_zona")
df_to = spark.table("bootcamp_de_valentin.gold.dim_tipo_operacion")
df_dataframe_api = (
    df
    .join(df_estado, df.caracteristicas_id == df_estado.caracteristica_id, "inner")
    .join(df_zona, df.zona_id == df_zona.zona_id, "inner")
    .join(df_to, df.tipo_operacion_id == df_to.tipo_operacion_id, "inner")
    .groupBy(df_zona.partido, df_to.tipo_operacion)
    .agg(
        count("*").alias("conteo"),
        avg(df.precio).alias("precio_promedio"),
        avg(df.precio_m2).alias("precio_m2_promedio")
    )
    .filter(col("conteo")>50)
    .orderBy(desc("precio_promedio"))
)
```
