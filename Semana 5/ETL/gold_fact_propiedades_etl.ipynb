```sql
-- ejercicio 2.5 etl
INSERT INTO bootcamp_de_valentin.gold.fact_propiedades_2 (
  row_hash,
  zona_id,  
  caracteristicas_id,  
  orientacion_id,  
  tipo_operacion_id,  
  fecha_id,  
  precio,
  precio_m2,
  expensas,
  m2_totales,
  m2_cubiertos,
  ambientes,
  url
)
SELECT 
  MD5(CONCAT_WS('|', ps.url, ps.precio)) AS row_hash,
  dz.zona_id,
  dc.caracteristicas_id,
  do.orientacion_id,
  dto.tipo_operacion_id,
  dt.fecha_id,
  ps.precio,
  ps.precio_por_m2 AS precio_m2,
  ps.expensas,
  ps.m2_totales,
  ps.m2_cubiertos,
  ps.ambientes,
  ps.url
FROM bootcamp_de_valentin.silver.propiedades_silver ps
  LEFT JOIN bootcamp_de_valentin.gold.dim_zona_2 dz ON 
    dz.partido = ps.partido 
    AND dz.region = ps.region
  LEFT JOIN bootcamp_de_valentin.gold.dim_caracteristicas_2 dc ON 
    dc.estado = COALESCE( ps.estado, 'sin especificar' )
    AND dc.cochera = COALESCE(ps.cochera, false)
  LEFT JOIN bootcamp_de_valentin.gold.dim_orientacion_2 do ON 
    do.orientacion = COALESCE( ps.orientacion, 'sin especificar')
  LEFT JOIN bootcamp_de_valentin.gold.dim_tiempo_2 dt ON 
    dt.fecha = ps.fecha_publicacion
  LEFT JOIN bootcamp_de_valentin.gold.dim_tipo_operacion_2 dto ON 
    dto.tipo_operacion = ps.tipo_operacion
    AND dto.moneda = ps.moneda
```
