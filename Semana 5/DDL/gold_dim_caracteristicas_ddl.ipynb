```sql
-- ejercicio 2.2 ddl
CREATE TABLE IF NOT EXISTS bootcamp_de_valentin.gold.dim_caracteristicas_2 (
  caracteristicas_id BIGINT GENERATED ALWAYS AS IDENTITY (START WITH 1 INCREMENT BY 1),
  estado STRING NOT NULL,
  cochera BOOLEAN,
  _created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP()
)
USING DELTA
TBLPROPERTIES ('delta.feature.allowColumnDefaults' = 'supported')
COMMENT 'Dimension de caracteristicas (estado + cochera) -- Gold -- BK: estado -- SK: caracteristicas_id'
```
