```sql
-- ejercicio 2.1
select
  c.cliente_id,
  c.nombre,
  c.ciudad,
  o.orden_id,
  o.producto_id,
  o.cantidad,
  o.fecha
from bootcamp.practice.clientes c
join bootcamp.practice.ordenes o ON c.cliente_id = o.cliente_id
```

```sql
-- ejercicio 2.2
select
  o.orden_id,
  o.fecha,
  p.nombre,
  p.categoria,
  p.precio,
  o.cantidad,
  p.precio * o.cantidad  as total

from bootcamp.practice.ordenes o
join bootcamp.practice.productos p on o.producto_id = p.producto_id 
;

-- select * from bootcamp.practice.ordenes
```

```sql
-- ejercicio 2.3
select
 c.nombre,
 c.ciudad,
 o.orden_id,
 o.fecha,
 p.nombre nombre_producto,
 p.categoria,
 p.precio,
 o.cantidad,
 p.precio * o.cantidad total
from bootcamp.practice.ordenes o
join bootcamp.practice.productos p on o.producto_id = p.producto_id
join bootcamp.practice.clientes c on c.cliente_id = o.cliente_id
```

```sql
-- Ejercicio 5.1
select
*
from bootcamp.practice.ordenes o
full outer join bootcamp.practice.clientes c on o.cliente_id = c.cliente_id
```

```sql
-- ejercicio 5.2
select
*
from bootcamp.practice.ordenes o
full outer join bootcamp.practice.clientes c on o.cliente_id = c.cliente_id
where 
  c.cliente_id is null  -- ordenes sin cliente
  or o.orden_id is null  -- cliente sin ordenes
```

```sql
-- ejercicio 6.1
select
  e.empleado_id,
  e.nombre nombre_empleado,
  e.cargo cargo_empleado,
  j.empleado_id jefe_id,
  j.nombre nombre_jefe,
  j.cargo cargo_jefe
from bootcamp.practice.empleados e
join bootcamp.practice.empleados j on e.jefe_id = j.empleado_id
```

```sql
-- ejercicio 6.2
select
  e.empleado_id,
  e.nombre nombre_empleado
from bootcamp.practice.empleados e
left join bootcamp.practice.empleados j on e.jefe_id = j.empleado_id
where j.empleado_id is null
```
