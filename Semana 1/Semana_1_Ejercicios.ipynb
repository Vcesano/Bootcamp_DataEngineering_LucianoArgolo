```sql
select
pickup_zip,
avg(trip_distance) as promedio_viaje
from samples.nyctaxi.trips
group by pickup_zip
```

```sql
select
pickup_zip,
count(*) conteo
from samples.nyctaxi.trips
group by pickup_zip
having conteo > 1000
```

```sql
/*
 Para zonas con más de 1,000 viajes, muestra: zona, cantidad de viajes, tarifa máxima, promedio y mínima. 
 Solo incluye viajes con tarifa mayor a $5 y distancia mayor a 0.
*/
select
pickup_zip,
count(*) cantidad_viajes,
max(fare_amount) max_precio,
round( avg(fare_amount), 2 ) avg_precio,
min(fare_amount) min_precio
from samples.nyctaxi.trips
where
  fare_amount > 5
  and trip_distance > 0
group by pickup_zip
having cantidad_viajes > 1000
```
